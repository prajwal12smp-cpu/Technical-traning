```python
#tree-hierarchial non-linear data structure
```

```python
class Tree:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
        
```

```python
#l=[5,10,4,3]
```

```python
r1=Tree(5)
l1=Tree(10)
l2=Tree(4)
l3=Tree(3)
```

```python
r1.right=l1
r1.left=l2
l2.left=l3
```

```python
r1.right.data
```

```python
l2.left.data
```

```python
def preorder(root):
    if root==None:
        return
    print(root.data)
    preorder(root.left)
    preorder(root.right)
```

```python
preorder(r1)
```

```python
def inorder(root):
    if root==None:
        return
    inorder(root.left)
    print(root.data)
    inorder(root.right)
```

```python
inorder(r1)
```

```python
def postorder(root):
    if root==None:
        return
    postorder(root.left)
    inorder(root.right)
    print(root.data)
```

```python
postorder(r1)
```
