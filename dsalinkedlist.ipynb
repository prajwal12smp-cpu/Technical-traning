```python
class Node:
    def __init__(self,data): #constructor
        self.data=data
        self.next=None
```

```python
n1=Node(10)
n2=Node(20)
n3=Node(30)
n4=Node(40)
n5=Node(50)
```

```python
cur=n1
while cur:
    print(cur.data)
    cur=cur.next
```

```python
#Find the sum of all nodes
```

```python
cur=n1
sum=0
while cur:
    sum+=cur.data
    cur=cur.next
    
print(sum)
    
```

```python
#Find the maximum element present in the node
```

```python
cur=n1
max_element=cur.data
while cur:
    cur.data>max_element
    max_element=cur.data
    cur=cur.next
    
print(max_element)
    
```
