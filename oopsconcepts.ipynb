```python
class bank:
    def __init__(self,Id,branch,ifsc,name):
        self.Id=Id
        self.branch=branch
        self.ifsc=ifsc
        self.name=name
    def bnk_info(self):
        msg=f'{self.Id}{self.branch}{self.ifsc}{self.name}'
        print(msg)
```
