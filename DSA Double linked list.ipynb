```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        self.prev=None
```

```python
n1=Node("A")
n2=Node("B")
n3=Node("C")
```

```python
n1.next=n2
n2.next=n3
n3.prev=n1
n3.prev=n2
```

```python
cur=n1
while cur:
    print(cur.data)
    cur=cur.next
```

```python
cur=n3
while cur:
    print(cur.data)
    cur=cur.prev
```
