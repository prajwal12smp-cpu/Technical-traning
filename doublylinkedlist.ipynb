```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        self.prev=None
```

```python
n1=Node("a")
n2=Node("b")
n3=Node("c")
```

```python
n1.next=n2
n2.next=n3
n2.prev=n1
n3.prev=n2
```

```python
cur=n1
while cur:
    print(cur.data)
    cur=cur.next
```

```python
cur=n3
while cur:
     print(cur.data)
     cur=cur.prev
    
```
