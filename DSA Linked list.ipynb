```python
class Node:
    def __init__(self,data): #constructor
        self.data=data
        self.next=None
```

```python
n1=Node(10)
n2=Node(20)
n3=Node(30)
```

```python
n1.data
```

```python
n2.data
```

```python
n3.data
```

```python
n1.next=n2
```

```python
n1.next
```

```python
n2.next=n3
```

```python
n2.next.data
```

```python
cur=n1
while cur:
    print(cur.data)
    cur=cur.next
```

```python
#find the sum of all nodes
```

```python
cur=n1
sum=0
while cur:
    sum+=cur.data
    cur=cur.next
print(sum)
```

```python
#find the maximum element present in the node
```

```python
cur=n1
max_element=cur.data
while cur:
    cur.data>max_element
    max_element=cur.data
    cur=cur.next
print(max_element)
```
