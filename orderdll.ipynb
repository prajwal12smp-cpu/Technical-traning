```python
class Order:
    def __init__(self,id,date,price,qty):
        self.id=id
        self.date=date
        self.price=price
        self.qty=qty
        self.next=None
        self.prev=None
    def order_info(self):
        msg=f'order id:{self.id}order date:{self.date}order price:{self.price}order qty:{self.qty}'
        print(msg)
```

```python
o1=Order(100,"10-10-2026",500,10)
```

```python
o2=Order(101,"11-10-2026",400,20)
```

```python
o3=Order(102,"12-10-2026",300,30)
```

```python
o1.order_info()
```

```python
o1.next=o2
o2.next=o3
o2.prev=o1
o3.prev=o2
```

```python
cur=o3
while cur:
    cur.order_info()
    cur=cur.prev
```

```python
cur=o1
o_id=int(input("Enter order id:"))
while cur:
    if cur.id==o_id:
        cur.order_info()
    cur=cur.next
```

```python
cur=o1
o_date=input("Enter the date:")
while cur:
    if cur.date==o_date:
        cur.order_info()
    cur=cur.next
```
